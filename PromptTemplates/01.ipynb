```python
from ..PromptTemplates import *

#方式1
# chatTemplate=ChatPromptTemplate.from_messages(
#     [
#         ('system','you are a english teacher,your name is {teacher_name}'),
#         ('user','I am {student_name},please tranlate "{word_input}" to english'),
#     ]
# )
#方式2
chatTemplate=ChatPromptTemplate(
    [
        ('system','you are a english teacher,your name is {teacher_name}'),
        ('user','I am {student_name},please tranlate "{word_input}" to english'),
    ]
)
word_input=input('Enter a word: ')
print('第一种格式化方法，返回ChatPromptValue类型\n')
prompt=chatTemplate.invoke({"teacher_name":'张三',"student_name":'hyz',"word_input":word_input})
print(type(prompt))
print(prompt)
print(len(prompt.messages))
```

```python
print('第二种格式化方法，返回str类型\n')
prompt=chatTemplate.format(teacher_name='张三',student_name='hyz',word_input=word_input)
print(type(prompt))
print(prompt)
print(len(prompt))
```

```python
print('第三种格式化方法，返回list类型\n')
prompt=chatTemplate.format_messages(teacher_name='张三',student_name='hyz',word_input=word_input)
print(type(prompt))
print(prompt)
print(len(prompt))
```
