```python
from ..memory import *
load_dotenv(override=True)
#必须关闭思考模式，deepseek默认为思考模式，思考模式不支持结构化输出
#因为deepseek底层使用tool_choice作为伪工具传递结构化输出，思考模式不支持tool_choice
DEEPSEEK_API_KEY=os.getenv('DEEPSEEK_API_KEY')
DEEPSEEK_BASE_URL=os.getenv('DEEPSEEK_BASE_URL')
model=init_chat_model(
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    model='deepseek-v4-flash',
    model_provider='deepseek',
    extra_body={"thinking":{"type":"disabled"}}
)
```

```python

db_url='postgresql://postgres:h2134468@localhost:5432/langchain_test?sslmode=disable'
#使用外部存储方式，即使进程结束了，也能保存消息，第二次运行同一段代码时，会添加同样的消息到上次历史中
#如果使用内存，第二次运行同一段代码时，不会添加同样的消息
#这里运行了两次本代码块
with PostgresSaver.from_conn_string(db_url) as checkpointer:
    checkpointer.setup()
    myagent=create_agent(
        model=model,
        checkpointer=checkpointer,
    )
    config={'configurable':{'thread_id':'test1'}}
    print("=" * 30, "-> 第一次调用 <-", "=" * 30)
    response1=myagent.invoke({
        'messages':HumanMessage('你好，我是hyz，请用十个字以内介绍你自己')
    },
    config=config,)
    rprint(response1['messages'])
    print("=" * 30, "-> 第二次调用 <-", "=" * 30)
    response2=myagent.invoke({
        'messages':HumanMessage('我是谁？然后介绍一下番茄炒蛋的做法，二十字内')
    },
    config=config,)
    rprint(response2['messages'])
```
