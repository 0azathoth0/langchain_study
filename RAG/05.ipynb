```python
from ..RAG import *
load_dotenv()
embed_url=os.getenv("SILI_FLOW_URL")
embed_key=os.getenv("SILI_FLOW_KEY")
#这里使用硅基流动中转平台免费的嵌入模型
embed_model=init_embeddings(
    model='BAAI/bge-m3',
    base_url=embed_url,
    api_key=embed_key,
    provider='openai',
)
text='床前明月光下一句是什么?'
#句子向量化
embeded_query=embed_model.embed_query(text=text)
print(embeded_query[:5])
print(len(embeded_query))
```

```python
#文档向量化
csv_path='../source/test.csv'
csv_loader=CSVLoader(csv_path)
texts=[data.page_content for data in csv_loader.load()]
embed_docs=embed_model.embed_documents(texts)
for i in range(len(texts)):
    print(f"{texts[i]}:\n{embed_docs[i][:3]}",end="\n\n")
```
