```python
from ..RAG import *
#使用token文本分割器，按照token进行分割
splitter=TokenTextSplitter(
    chunk_size=33,
    chunk_overlap=0,
    encoding_name='cl100k_base'#使用的编码器，用于将文本序列化为token,在openai_public.py中有
)
text = "人工智能是一个强大的开发框架。它支持多种语言模型和工具链。人工智能是指通过计算机程序模拟人类智能的一门科学。自20世纪50年代诞生以来，人工智能经历了多次起伏。"
text_list=splitter.split_text(text)
print(f"原始文本被分割成了 {len(text_list)} 个块:")
for i, chunk in enumerate(text_list):
    print(f"块 {i+1}: 长度：{len(chunk)} 内容：{chunk}")
    print("-" * 50)
```

```python
#调用分割器的序列化方法from_tiktoken_encoder
splitter2=CharacterTextSplitter.from_tiktoken_encoder(
    chunk_size=18,
    chunk_overlap=0,
    encoding_name='o200k_base',
    separator='。',
    keep_separator=False,#Chunk中是否保留分割符
)
text_list2=splitter2.split_text(text)
for i, chunk in enumerate(text_list2):
    print(f"块 {i+1}: 长度：{len(chunk)} 内容：{chunk}")
    print("-" * 50)
```
