```python
from RAG import *
#这些各种加载器都继承自BaseLoader,实现其中的load方法，将不同的数据源统一加载为Document对象
#文档加载器，以txt加载器为例
txt_path= '../source/test.txt'
text_loader=TextLoader(file_path=txt_path,encoding='utf-8')
document=text_loader.load()
print(document)
print(type(document))
```

```python
#csv加载器
csv_path='../source/test.csv'
csv_loader=CSVLoader(csv_path)
data=csv_loader.load()
print(type(data)) # <class 'list'>
print(type(data[0])) # <class 'langchain_core.documents.base.Document'>
print(len(data)) #
print(data[0].page_content)
```

```python
#使用jq和jsonLoader
json_loader=JSONLoader(
    file_path='../source/test.json',
    jq_schema='.',#提取所有字段,需要安装jq库
    text_content=False,#保持原始 JSON 结构，将提取的数据转换为JSON字符串存入page_content
)
docs=json_loader.load()
rprint(docs)
```
