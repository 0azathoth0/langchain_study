```python
from ..RAG import *
#文档切割器父类TextSplitter,里面包含了各种切分方法
#主要调用链为split_documents调用create_documents，create_documents再调用split_text，这些方法由具体的子类实现
#split_documents(documents)
# -> create_documents(texts, metadatas=metadatas)
# -> split_text(text)
#创建splitter时的参数包含，chunk_size每个切块的最大字符数量，chunk_overlap相邻切块间可重叠字符数量用于保障语义完整，length_function计算切块长度的函数一般默认使用父类的，add_start_index在元数据中包含chunk的起始索引
#CharacterTextSplitter字符串分割器
#separator按照你指定的分隔符进行分割，默认为\n\n
#如果存在separator(不为空字符串或默认值)，则先按照separator分割，再考虑chunk_size和chunk_overlap，否则即使chunk_size和chunk_overlap超了，也不会分割
text='你好，hyz，我是AI助手，能帮你做一些事'
splitter=CharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0,
    separator='，',
)
list_str=splitter.split_text(text)
print(list_str)
for i,chunk in enumerate(list_str):
    print(f"块 {i+1}:长度：{len(chunk)}")
    print(chunk)
    print("-" * 50)
```

```python
#RecursiveCharacterTextSplitter递归字符串分割器，最常用
#参数同父类,它的separator中包含了默认值["\n\n", "\n", " ", ""]或者自己传入的separators，优先按照自然文本边界进行切分
splitter_2=RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0,
    add_start_index=True,
    # separators=['\n\n','\n','。','，',' ','']#自定义separators
)
text2="LangChain框架特性\n\n多模型集成(GPT/Claude)\n\记忆管理功能\n链式调用设计。文档分析场景示例：需要处理PDF/Word等格式。"
paragraph=splitter_2.split_text(text2)
for i,chunk in enumerate(paragraph):
    print(f"块{i + 1},长度：{len(chunk)}")
    print(chunk)
    print('-' * 50)
```

```python
#分割过程，递归分割，先按照顺序使用\n\n分割，然后\n分割，然后按照空格分割，最后再按照""字符级进行分割
#如果分割完后，保留的字符数小于chunk_size则保留，否则跳转到下一个分割符进行分割
text2_list=[text2]
paragraphs=splitter_2.create_documents(text2_list)
for para in paragraphs:
    print(para)
```

```python
txt_path= '../source/test.txt'
text_loader=TextLoader(file_path=txt_path,encoding='utf-8')
documents=text_loader.load()
#使用split_documents，传入Document对象列表，返回切割后的Document对象列表
split_docs=splitter_2.split_documents(documents)
for doc in split_docs:
    print(doc)
    print('-'*10)
```
