```python

from  ..message import *
load_dotenv(override=True)
DEEPSEEK_API_KEY=os.getenv('DEEPSEEK_API_KEY')
DEEPSEEK_BASE_URL=os.getenv('DEEPSEEK_BASE_URL')

model=init_chat_model(
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    model='deepseek-v4-flash',
    model_provider='deepseek',
    extra_body={"thinking":{"type":"enabled"}}
)
messages=[
    {'role':"system",'content':'你是一个历史学家'},
    {'role':"user",'content':'苏联什么时候解体的，输出时间格式yyyy-mm-dd'},
]
response=model.invoke(messages)
print(type(response))
# print(response.content,end='', flush=True)
# print("",end='\n')
# print(response.additional_kwargs.get("reasoning_content"),end='', flush=True)
print("="*5,'response.content',"="*5,end='\n')
print(response.content)
print("="*5,'response.content_blocks',"="*5,end='\n')
print(response.content_blocks)
```

```python
messages.append({'role':"assistant",'content':'1991-12-26'})
messages.append({'role':"user",'content':'我刚刚问了什么问题？'})
for chunk in model.stream(messages):
    print(chunk.content,end="",flush=True)
```
